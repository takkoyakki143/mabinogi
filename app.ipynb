```python
player_1 = {
    "name": "문어야끼",
    "combat_power": 106586,
    "job": "화염술사",
    "magic_resistance": 4716,
    "is_sub": False,
    "is_newbie": False,
    "play_device": "PC"
}
```

```python
player_2 = {
    "name": "애샷추",
    "combat_power": 103920,
    "job": "기사",
    "magic_resistance": 5200,
    "is_sub": False,
    "is_newbie" : False,
    "play_device": "PC"
}
```

```python
players = [player_1, player_2]
```

```python
print(players[0]["name"])
print(players[1]["job"])
```

```python
combat_powers = []

for player in players:
    combat_powers.append(player["combat_power"])

print(combat_powers)
```

```python
average_combat_power = sum(combat_powers) / len(combat_powers)

print(average_combat_power)
```

```python
player_3 = {
    "name": "다람룡",
    "combat_power": 101651,
    "job": "악사",
    "magic_resistance": 4512,
    "is_sub": False,
    "is_newbie" : False,
    "play_device": "PC"
}
```

```python
players = [player_1, player_2, player_3]
```

```python
combat_powers = []

for player in players:
    combat_powers.append(player["combat_power"])

print(combat_powers)
```

```python
average_combat_power = sum(combat_powers) / len(combat_powers)

print(round(average_combat_power))
```

```python
magic_resistances = []

for player in players:
    magic_resistances.append(player["magic_resistance"])

print(magic_resistances)
```

```python
magic_resistance = 5500

if magic_resistance >= 5400:
    print("마도저항 보정: +10%")
elif magic_resistance >= 4400:
    print("마도저항 보정: +5%")
else:
    print("마도저항 보정: 0%")
```

```python
combat_power = 106586
magic_resistance = 4716

if magic_resistance >= 5400:
    mr_bonus = 0.10
elif magic_resistance >= 4400:
    mr_bonus = 0.05
else:
    mr_bonus = 0

adjusted_combat_power = combat_power * (1 + mr_bonus)

print(round(adjusted_combat_power))
```

```python
for player in players:
    combat_power = player["combat_power"]
    magic_resistance = player["magic_resistance"]

    if magic_resistance >= 5400:
        mr_bonus = 0.10
    elif magic_resistance >= 4400:
        mr_bonus = 0.05
    else:
        mr_bonus = 0

    adjusted_combat_power = combat_power * (1 + mr_bonus)

    print(player["name"], round(adjusted_combat_power))
```

```python
def get_mr_bonus(magic_resistance):
    if magic_resistance >= 6200:
        return 0.15
    elif magic_resistance >= 5400:
        return 0.10
    elif magic_resistance >= 4400:
        return 0.05
    elif magic_resistance >= 2500:
        return 0
    elif magic_resistance >= 2400:
        return -0.10
    elif magic_resistance >= 2300:
        return -0.15
    elif magic_resistance >= 2200:
        return -0.30
    elif magic_resistance >= 2100:
        return -0.40
    elif magic_resistance >= 2000:
        return -0.45
    elif magic_resistance >= 1900:
        return -0.55
    elif magic_resistance >= 1800:
        return -0.60
    elif magic_resistance >= 1700:
        return -0.70
    elif magic_resistance >= 1600:
        return -0.80
    else:
        return -0.90

print(get_mr_bonus(6500))
print(get_mr_bonus(1500))
```

```python
def get_adjusted_combat_power(combat_power, magic_resistance):
    mr_bonus = get_mr_bonus(magic_resistance)
    adjusted_combat_power = combat_power * (1 + mr_bonus)

    return round(adjusted_combat_power, 2)

get_adjusted_combat_power(106586, 4716)
```

```python
team_job_bonus = {
    "사제": 0.2,
    "전사": 0.2,
    "수도사": 0.2,
    "기사": 0.2,
    "빙결술사": 0.2,
    "힐러": 0.3
}

def get_job_bonus(job):
    return job_bonus.get(job, 0)

print(get_job_bonus("기사"))
print(get_job_bonus("화염술사"))
```

```python
special_player_bonus = {
    "아키이즈": 0.7,
    "레키나": 0.15
}
```

```python
def get_special_player_bonus(name):
    return special_player_bonus.get(name, 0)

print(get_special_player_bonus("문어야끼"))
print(get_special_player_bonus("아키이즈"))
```

```python
def get_team_average(team):
    adjusted_powers = []

    for player in team:
        adjusted_powers.append(player["adjusted_combat_power"])

    average_power = sum(adjusted_powers) / len(adjusted_powers)

    return round(average_power, 2)
```

```python
team_a = players

print(get_team_average(team_a))
```

```python
def get_team_job_bonus(team):
    jobs = []

    for player in team:
        jobs.append(player["job"])

    total_bonus = 0

    for job in team_job_bonus:
        if job == "빙결술사":
            continue

        if job in jobs:
            total_bonus += team_job_bonus[job]

    ice_count = jobs.count("빙결술사")

    if ice_count >= 2:
        total_bonus += 0.4
    elif ice_count == 1:
        total_bonus += 0.2

    return total_bonus

print(get_team_job_bonus(team_a))
```

```python
def get_team_special_bonus(team):
    total_bonus = 0

    for player in team:
        name = player["name"]

        if name in special_player_bonus:
            total_bonus += special_player_bonus[name]

    return total_bonus

print(get_team_special_bonus(team_a))
```

```python
def get_team_player_penalty(team):
    total_penalty = 0

    for player in team:
        if player["is_sub"]:
            total_penalty -= 0.3

        if player["is_newbie"]:
            total_penalty -= 0.7

    return total_penalty

print(get_team_player_penalty(team_a))
```

```python
def get_team_score(team):
    average_power = get_team_average(team)
    job_bonus = get_team_job_bonus(team)
    special_bonus = get_team_special_bonus(team)
    player_penalty = get_team_player_penalty(team)

    team_score = average_power + job_bonus + special_bonus + player_penalty

    return round(team_score, 2)

print(get_team_score(team_a))
```

```python
def print_team_result(team_name, team):
    average_power = get_team_average(team)
    job_bonus = get_team_job_bonus(team)
    special_bonus = get_team_special_bonus(team)
    player_penalty = get_team_player_penalty(team)
    team_score = get_team_score(team)

    print(f"[{team_name}]")
    print(f"마도저항 반영 평균 전투력 : {average_power:.2f}")
    print(f"직업 보너스         : {job_bonus:+.2f}")
    print(f"특수 캐릭터 보너스  : {special_bonus:+.2f}")
    print(f"부캐/뉴비 보정      : {player_penalty:+.2f}")
    print(f"최종 팀 점수        : {team_score:.2f}")


print_team_result("A팀", team_a)
print()
print_team_result("B팀", team_b)
```

```python
from itertools import combinations

team_combinations = list(combinations(players_2, 3))

print(len(team_combinations))
```

```python
for team in team_combinations:
    names = []

    for player in team:
        names.append(player["name"])

    print(names)
```

```python
team_a = team_combinations[0]

team_b = []

for player in players_2:
    if player not in team_a:
        team_b.append(player)

print("A팀")

for player in team_a:
    print(player["name"])

print()

print("B팀")

for player in team_b:
    print(player["name"])
```

```python
results = []

for team_a in team_combinations:
    team_b = []

    for player in players_2:
        if player not in team_a:
            team_b.append(player)

    team_a_score = get_team_score(team_a)
    team_b_score = get_team_score(team_b)

    score_difference = abs(team_a_score - team_b_score)

    results.append({
        "team_a": team_a,
        "team_b": team_b,
        "team_a_score": team_a_score,
        "team_b_score": team_b_score,
        "score_difference": score_difference
    })

print(len(results))
```

```python
sorted_results = sorted(
    results,
    key=lambda result: result["score_difference"]
)

for result in sorted_results:
    print(result["score_difference"])
```

```python
unique_results = []

first_player = players_2[0]

for result in sorted_results:
    if first_player in result["team_a"]:
        unique_results.append(result)

```

```python
top_3_results = unique_results[:3]

for index, result in enumerate(top_3_results, start=1):
    print(f"추천안 {index}")

    print("A팀:")
    for player in result["team_a"]:
        print("-", player["name"])

    print("B팀:")
    for player in result["team_b"]:
        print("-", player["name"])

    print("A팀 점수:", f'{result["team_a_score"]:.2f}')
    print("B팀 점수:", f'{result["team_b_score"]:.2f}')
    print("점수 차이:", f'{result["score_difference"]:.2f}')
    print()
```
